**Apache Spark**
unified, open-source compute engine designed for large-scale, parallel data processing.

**Key capabilities and advantages of Spark include:**
**Unified Processing:** Spark supports multiple data processing styles within one framework, including *batch processing*, *stream processing*, *machine learning* (e.g., classification, regression), and *graph processing*.
**uperior Speed:** It is noted to be up to 100 times faster than *Hadoop MapReduce* because it utilizes **in-memory processing**, storing intermediate results in the RAM of nodes rather than writing them to disk.
**Multi-Language Support:**
    The core of Spark is written in *Scala*, which is considered its default language.
    It provides strong support for *SQL* (the most common interface), *Python* (via *PySpark*), *Java*, and *R* for statistical computing.

**Apache Spark architecture**
using a real-life analogy involving a wealthy person needing to count money in multiple warehouses. The analogy maps to the core components of a Spark application:

* **Client**: The program or person that initiates the requirement and submits the code.
* **Driver**: The planner that interprets the code, converts it to Java byte code, manages the *SparkContext*, and coordinates the job execution.
* **Cluster Manager**: The coordinator that identifies and provides available resources (*worker nodes*) for the job. Examples include *Standalone*, *Hadoop YARN*, *Apache Mesos*, and *Kubernetes*.
* **Executors**: The worker resources that perform the actual processing tasks assigned to them by the driver.

The video details the **workflow** where the driver requests executors from the cluster manager, sends byte code and instructions to those executors, and finally consolidates the results to send back to the client. It also clarifies that these components run on specific **nodes**, where a *Driver Node* hosts the driver and *Worker Nodes* host the executors.

**Apache Spark APIs:** 
used for processing big data, categorizing them from low-level foundations to modern structured interfaces.

### **Low-Level APIs**
* **RDDs (Resilient Distributed Datasets):** The foundational API designed for fault-tolerant, distributed data processing across multiple nodes.
* **Distributed Variables:** Programming variables shared across all nodes in a cluster to facilitate communication.

### **Structured APIs**
* These modern APIs are the industry standard, presenting data as rows and columns similar to tables.
* **DataFrames:** The primary API for programmatic interaction using languages like *PySpark*, *Scala*, and *Java*.
* **SQL:** Allows developers to interact with data using standard SQL syntax.
* **Datasets:** Similar to DataFrames but provide type safety for objects, though they are less commonly used today.

### **High-Level Libraries**
* Spark offers specialized libraries built on the structured APIs for advanced tasks, including:
    * **Structured Streaming:** For processing real-time streaming data.
    * **MLlib & Spark R:** For machine learning and heavy statistical computations.
    * **GraphX:** For graph processing.

In this series, the focus will remain on **DataFrames** and **Spark SQL** as the primary tools for modern development.
